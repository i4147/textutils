# 1. Simple Jawi

HarfBuzz reads the font's own OpenType tables and gives positioned glyphs
back. `Layout` then sets the position of each glyph.

```python
from textsynth import Layout, Synthesizer, render
from nbutil import FONT, sheet

TEXT = "سايا ممباچ سورت خبر"
render(TEXT, FONT, size=72)
```

## Letter spacing

`tracking` changes the space between the glyphs. The joins stay correct.

```python
rows = [
    (
        f"tracking = {t}",
        Synthesizer(font=FONT, layout=Layout(size=64, tracking=t, supersample=2))(TEXT),
    )
    for t in (-2, -1, 0, 2, 5)
]
sheet(rows, gutter=140)
```
